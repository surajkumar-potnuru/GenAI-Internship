```python
# scolarship selection
marks = [96, 81, 55, 74, 67, 92]
for mark in marks:
  if mark>=90:
    print("Full scholarship")
  elif mark > 75 and mark < 89:
    print("Half scholarship")
  elif mark > 60 and mark <= 74:
    print("Training program")
  else:
    print("Better Luck next time.")
```

```python
#Shopping cart
prices = [250, 180, 450, 120, 600]
total = 0
highest = float('-inf')
lowest = float('inf')
for price in prices:
  total += price
  if price > highest:
    highest = price
  if price < lowest:
    lowest = price
print(total)
print(highest)
print(lowest)
```
