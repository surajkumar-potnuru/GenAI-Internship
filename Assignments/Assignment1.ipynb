```python
#Shopping cart
prices = [250, 180, 450, 120, 600]
total = 0
highest = float('-inf')
lowest = float('inf')
for price in prices:
  total += price
  if price > highest:
    highest = price
  if price < lowest:
    lowest = price
print(total)
print(highest)
print(lowest)
```
