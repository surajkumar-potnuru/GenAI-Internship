```python

# Problem 1: Student Report Card
students = {"Rahul":91,"Ayesha":84,"Kiran":67,"Ali":45}
passed = 0
for student,marks in students.items():
  print(student,marks)
  if marks >= 40:
    print("Pass")
    passed += 1
  else:
    print("Fail")
print("No of students passed",passed)
```
