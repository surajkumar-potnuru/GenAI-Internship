```python
users={"suraj":"hello","admin":"python123"}
attempts=0
while attempts<3:
    username=input("Enter username: ")
    password=input("Enter password")
    if username in users:
        if users[username] == password:
            print("login successful")
            break
        else:
            print("Enter correct password")
    else:
        print("user not found")
    attempts+=1
if attempts == 3:
    print("Acount locked")
    
```

```python
students={"Rahul":91,"ayesha":84,"kiran":67,"ali":45}
print(students.items())
passcount=0
failcount=0
for name,marks in students.items():
    if marks >= 45:
        print(name,"pass")
        passcount+=1
    else:
        print(name,"fail")
        failcount+=1
print("No of students pass: ",passcount)
print("No of students fail",failcount)
    
```
